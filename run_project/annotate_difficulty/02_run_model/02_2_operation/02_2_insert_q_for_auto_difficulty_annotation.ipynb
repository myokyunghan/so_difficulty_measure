```python
import lib.annotation.tools.Sample_Insert as si
from datetime import datetime, timedelta
```

```python
def return_date_list(stday) :
    start_date = datetime(2020, 11, 1)
    end_date = datetime(2025, 12, 1)

    # 결과 날짜를 저장할 리스트
    dates = []

    # 현재 날짜를 시작 날짜로 설정
    current_date = start_date

    while current_date <= end_date:
        chk_date = (datetime(current_date.year, current_date.month, 1) + timedelta(days=stday-1))
        if chk_date <= end_date :
            dates.append([
                            (datetime(current_date.year, current_date.month, 1) + timedelta(days=stday-1)).strftime("%Y-%m-%d"), 
                            (datetime(current_date.year, current_date.month, 1) + timedelta(days=stday)).strftime("%Y-%m-%d")
                        ])
            # 다음 달로 이동
            year = current_date.year
            month = current_date.month + 1
            if month > 12:
                month = 1
                year += 1
            current_date = datetime(year, month, 1)
        else :
            break

    return dates
```

```python
from datetime import datetime, timedelta

def return_from_to_date_list():
    start_date = datetime(2020, 11, 1)
    end_date = datetime(2025, 12, 1)

    dates = []
    current_date = start_date

    while current_date <= end_date:
        next_date = current_date + timedelta(days=1)

        dates.append([
            current_date.strftime("%Y-%m-%d"),
            next_date.strftime("%Y-%m-%d")
        ])

        current_date = next_date

    return dates
```
