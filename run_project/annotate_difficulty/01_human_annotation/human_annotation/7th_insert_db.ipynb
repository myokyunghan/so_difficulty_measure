```python
import pandas as pd
import os
import config.config as conf
import psycopg2
import psycopg2.extras
import sqlalchemy as sa
from sqlalchemy import create_engine
```

```python
file_list = os.listdir('.')
```

```python
tar_list = [x for x in file_list if x.endswith('snapshop2_sample.csv')]
```

```python
tot_df = pd.DataFrame()
for tar in tar_list:
    tot_df = pd.concat([tot_df, pd.read_csv(tar)], axis = 0)
```

```python
q_list = [(x,) for x in tot_df['id']]
```

```python
engine_str = "postgresql://{}:{}@{}/{}".format(conf.database_user['user'],conf.database_user['password'],conf.database_user['host'],conf.database_user['dbname'])
engine = sa.create_engine(engine_str, client_encoding='utf8')
conn = engine.raw_connection()
cursor = conn.cursor()

try:
    sql = f'INSERT INTO {conf.SCHEMA_NAME}.tt_posts_difficulty_annotated  VALUES %s'
    psycopg2.extras.execute_values(cursor, sql, q_list, template=None, page_size=100)
    conn.commit()
except Exception as e:
    print('Error : ', e)
finally:
    conn.commit()
    cursor.close()
    conn.close()            
```
